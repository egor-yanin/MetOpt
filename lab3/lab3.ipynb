```python
import sympy as sp
x, y = sp.symbols("x y", real=True)
lam = sp.symbols("lam", real=True)
f = x**2 + y**2
constraints = [x + y - 4]
L = f + lam * constraints[0]
equations = [sp.diff(L, v) for v in (x, y)] + constraints
for sol in sp.solve(equations, [x, y, lam], dict=True):
    pt = {x: sol[x], y: sol[y]}
    print(sol, sp.simplify(constraints[0].subs(pt)), sp.simplify(f.subs(pt)))
```
